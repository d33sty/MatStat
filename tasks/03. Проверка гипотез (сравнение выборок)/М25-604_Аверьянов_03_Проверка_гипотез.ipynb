# Решение

В первую очередь импортируем необходимые модули и занесем данные из таблицы 1 в массивы numpy.

```python
import numpy as np
import scipy.stats as stats

# данные из таблицы 1 (износ протектора, мм)
standard = np.array([
    3.099999905, 4.099999905, 4.699999809, 4.800000191, 4.300000191,
    4.699999809, 4.599999905, 3.599999905, 6.000000000, 3.599999905,
    4.199999809, 4.900000095, 4.300000191, 5.699999809, 3.200000048,
    3.799999952, 5.000000000, 4.500000000, 5.800000191, 5.300000191,
])

new = np.array([
    2.900000095, 2.599999905, 5.000000000, 3.700000048, 4.599999905,
    4.099999905, 3.099999905, 4.599999905, 5.300000191, 3.500000000,
    4.300000191, 5.099999905, 3.099999905, 5.099999905, 3.799999952,
    2.299999952, 4.000000000, 3.900000095, 4.099999905, 3.700000048,
])

n = len(standard)
print(f"Количество грузовиков: n = {n}")
```

## 1. Постановка гипотезы

Данный эксперимент является **парным** (зависимые выборки), поскольку оба вида шин испытывались одновременно на одном и том же грузовике. Каждый грузовик образует пару наблюдений (стандартная шина, новая шина). Это исключает влияние различий между грузовиками (масса, стиль вождения, маршрут) на результат.

Исследователь выбрал парный дизайн вместо независимого именно для того, чтобы **устранить межгрупповую изменчивость**: в независимом эксперименте разброс от грузовика к грузовику замаскировал бы разницу между шинами.

Введем случайную величину разностей для каждой пары:
$$D_i = X_{\text{ст},i} - X_{\text{нов},i}$$

**Нулевая гипотеза** $H_0$: $\mu_D = 0$ — в среднем износ одинаков.

**Альтернативная гипотеза** $H_1$: $\mu_D \ne 0$ — средний износ различается (двусторонняя проверка).

Уровень значимости: $\alpha = 0{,}05$.

## 2. Предварительный анализ данных

```python
print("Стандартная шина:")
print(f"  Среднее: x_ст = {np.mean(standard):.4f} мм")
print(f"  СКО:     s_ст = {np.std(standard, ddof=1):.4f} мм")
print()
print("Новая шина:")
print(f"  Среднее: x_н  = {np.mean(new):.4f} мм")
print(f"  СКО:     s_н  = {np.std(new, ddof=1):.4f} мм")
```

## 3. Анализ парных разностей

Вычислим разности $d_i = x_{\text{ст},i} - x_{\text{нов},i}$ для каждого грузовика, затем найдем их выборочное среднее и стандартное отклонение:

$$\bar{d} = \frac{1}{n}\sum_{i=1}^{n} d_i, \qquad
s_d = \sqrt{\frac{\sum_{i=1}^{n}(d_i - \bar{d})^2}{n-1}}, \qquad
s_{\bar{d}} = \frac{s_d}{\sqrt{n}}$$

```python
d = standard - new

d_mean  = np.mean(d)
d_std   = np.std(d, ddof=1)
d_se    = d_std / np.sqrt(n)

print("Статистические характеристики разностей:")
print(f"  Среднее:           d_mean = {d_mean:.4f} мм")
print(f"  СКО разностей:     s_d    = {d_std:.4f} мм")
print(f"  СКО среднего:      s_d_sr = {d_se:.4f} мм")
```

## 4. Проверка гипотезы — парный t-критерий

При $H_0$ наблюдаемая статистика
$$t = \frac{\bar{d}}{s_{\bar{d}}}$$
имеет распределение Стьюдента с $df = n - 1$ степенями свободы.

Критическое значение при двустороннем критерии ($\alpha/2 = 0{,}025$ в каждом хвосте):
$$t_{\text{кр}} = t_{0{,}975,\, df}$$

$H_0$ отвергается, если $|t| > t_{\text{кр}}$.

```python
t_stat     = d_mean / d_se
df         = n - 1
t_critical = stats.t.ppf(0.975, df=df)
p_value    = 2 * (1 - stats.t.cdf(abs(t_stat), df=df))

print(f"t-статистика:            t     = {t_stat:.4f}")
print(f"Степени свободы:         df    = {df}")
print(f"Критическое значение:    t_кр  = {t_critical:.4f}")
print(f"p-значение:              p     = {p_value:.6f}")
```

## 5. Доверительный интервал для средней разности

95%-й доверительный интервал для $\mu_D$:
$$\bar{d} \pm t_{\text{кр}} \cdot s_{\bar{d}}$$

```python
ci_lower = d_mean - t_critical * d_se
ci_upper = d_mean + t_critical * d_se

print(f"95% доверительный интервал для средней разности:")
print(f"  [{ci_lower:.4f};  {ci_upper:.4f}] мм")
```

## 6. Выводы

```python
print(f"Средняя разность износа (стандартная - новая): d = {d_mean:.4f} мм")
print(f"95% ДИ: [{ci_lower:.4f}; {ci_upper:.4f}] мм")
print()

if abs(t_stat) > t_critical:
    print(f"|t| = {abs(t_stat):.4f} > t_кр = {t_critical:.4f}")
    print("Нулевая гипотеза ОТВЕРГАЕТСЯ на уровне значимости alpha = 0.05")
    if d_mean > 0:
        print("Вывод: износ стандартной шины статистически значимо БОЛЬШЕ износа новой шины.")
    else:
        print("Вывод: износ новой шины статистически значимо БОЛЬШЕ износа стандартной шины.")
else:
    print(f"|t| = {abs(t_stat):.4f} <= t_кр = {t_critical:.4f}")
    print("Нулевая гипотеза НЕ отвергается на уровне значимости alpha = 0.05")
    print("Статистически значимой разницы в износе шин не обнаружено.")
```

**Почему парный эксперимент, а не независимый?**

В независимом эксперименте каждый грузовик попал бы только в одну из групп. Разброс, обусловленный различиями между грузовиками (нагрузка, маршрут, водитель), вошёл бы в оценку дисперсии и «замаскировал» бы реальную разницу между шинами. В парном дизайне каждый грузовик служит собственным контролем: межиндивидуальная изменчивость полностью исключается из анализа, а тест работает только с разностями внутри каждой пары — это повышает мощность критерия.
